# DevScope Notebook 01: Multi-Year ETL Pipeline & Data Cleaning

This notebook demonstrates the end-to-end Extract, Clean, Transform, and Load (ETL) pipeline for the **DevScope Developer Career Intelligence Platform**.

### Stakeholder Business Case
**GlobalTech Enterprises** requires clean, standardized, multi-year developer survey data (2021–2024) and job postings benchmarks to optimize compensation structures and tech stack investments.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

from etl.extract import extract_survey_datasets, extract_job_postings
from etl.clean import clean_all_surveys, clean_job_postings
from etl.transform import transform_survey_features, compute_executive_kpis
from etl.load import load_data_to_database
```

## Step 1: Extraction & Schema Validation

```python
raw_surveys = extract_survey_datasets()
raw_jobs = extract_job_postings()
print(f"Extracted {len(raw_surveys)} survey years and {len(raw_jobs)} job postings.")
```

## Step 2: Data Cleaning & Missing Value Protocols

```python
cleaned_surveys = clean_all_surveys(raw_surveys)
cleaned_jobs = clean_job_postings(raw_jobs)
cleaned_surveys.head()
```

## Step 3: Feature Engineering & Executive KPIs

```python
transformed = transform_survey_features(cleaned_surveys)
kpis = compute_executive_kpis(transformed)
print("Executive KPIs:", kpis)
```

## Step 4: Relational Database Load

```python
load_data_to_database()
print("Database population completed successfully!")
```
